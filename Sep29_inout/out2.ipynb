```python
# format (포맷) : 원하는 서식(형태)의 문자열로 만들어서 출력
# 변하는 데이터를 깔끔하고 사용자가 보기 편한 하나의 문자열로 완성
# 전체적인 틀은 유지하고 변경되는 값 자리만 뚫어두기
```

```python
# print('서식'%(값, 값...))
# %d : 정수, %f : 실수 %s : 문자열 
# 정수 : 소수점이 없는 숫자, 실수 : 소수점이 있는 숫자
print('%d년 %02d월 %d일'%(26, 9, 29)) # %와 d 사이에 02는 자릿수를 의미 
# 문자열 표현:'문자열' or "문자열"
# hello!! i'm 이름 - 중간에 ' 들어가 있으면 " 를 써주는게 좋다.(오류 방지)
print("hello!! i'm %s"%'김종현')
```

```python
print('%d'%12345)
print('%3d'%12345)
print('%03d'%12345)
print('%09d'%12345)
print('%9d'%12345)
```

```python
print('%f'%3.1415)
print('%.2f'%3.1415)
print('%.2f'%3.1465) # 소수점 2자리 까지만 입력, 뒤에는 반올림
```

```python
# format()
# 2 * 2 = 4
print('{} * {} =  {}'.format(2,2,4))
print('{} * {} =  {}'.format(2,3,6))
```

```python
# f'' : 변수에 저장된 값을 꺼내와 {}에 넣어준다 (값 직접 지정해도 됨) f 는 format 약자
# 변수 : 하나의(하나만) 값을 저장하는 공간
name = '김종현'
age = 10
# {변수명}, {값} : 출력할 구조가 한눈에 보임
print(f'안녕 내 이름은 {name}이고 나이는 {age}살이고 직업은{"초등학생"}이야')
```

```python
# 동일한 틀에 데이터만 계속 바꿔 들어가는 시스템을 만들 때 포맷팅을 사용한다.(로그 찍기 등)
```

```python
print(f"{7:3}") # :3 세자리
print(f"{7:03}") # :03 :0을 채워서 세자리

money = 1234567
print(f"{money:,}원") #:, 세자리마다 컴마

# < : 왼쪽 정렬 > : 오른쪽 정렬 ^ : 가운데 정렬
print(f"{'번호':<6}{'이름':<10}{'점수':>6}") # < - 글자를 왼쪽으로 밀어줌 / > - 반대 정렬
print(f"{1:<6}{'홍길동':<10}{95:>6}") # 숫자는 그 옆으로 자릿수 / ^ : 가운데 정렬
print(f"{2:<6}{'김철수':<10}{87:>6}")
```

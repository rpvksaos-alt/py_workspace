```python
# 주석/코멘트 : 프로그램의 실행에 영향을 주지 않는 설명/메모
#            다른 사람이 코드를 쉽게 이해할 수 있도록 작성
# 드래그 후 ctrl + / - 주석 달고 풀기
```

```python
# 출력문
#print (값) : 괄호 안의 값을 받아서 화면 (터미널)에 출력 후 줄바꿈 (커서가 내려감)
print(10)
print(100)
```

```python
# print (값, 값, 값, 값) : 값 여러개를 한번에 출력
print (10, 20, 30)
print (30, 40, 50)
```

```python
# print(값, 값, ..., sep='구분자' ) : 값 사이에 공백이 아닌 다른 구분자를 넣어주고 싶다.
print(10, 20, 30, sep='&&')
```

```python
#print(값, end = "프린트가 끝나고 나서 마지막에 출력할 문자열')
print(10, end='+++++')
print(100)
```

```python
# \n : 줄바꿈(라인 피드)
print(10, end='\n\n\n')
print(11111)
```

```python
print(10, 20, 30, end = '', sep = '-')
print(50, 60)
```

```python
'''자주 발생하는 예외 
예외	                    발생 상황
ValueError	             "abc"를 정수로 변환
ZeroDivisionError	      0으로 나누기
IndexError	              리스트에 없는 인덱스 접근
KeyError	              딕셔너리에 없는 키 접근
TypeError	              지원하지 않는 타입끼리 연산
FileNotFoundError	      존재하지 않는 파일 접근'''
```

```python
numbers = [ 10,20,30 ]
try:
    10 / 0 # ZeroDivisionError
    print(numbers[5])
except IndexError:
    print('해당 위치(인덱스)의 값 없음')
```

```python
# 다양한 예외가 발생할 수 있음
# 각각의 예외 타입별로 다르게 예외 처리
try:
    str_num = input('숫자 입력')
    print(100/int(str_num))
except ValueError:
    print('숫자만 입력하세요')
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다.')
except:
    print('예외 발생')
    
print('정상 종료')
```

```python
my_str = '파이썬은 재밌다. 매일매일 파이썬 공부 하고 싶다. 파이썬 최고'
str_pos_list = []
position = 0

while True:
    try:
        position = my_str.index('파이썬', position)
        str_pos_list.append(position)
        position += 1
    except:
        break

print('파이썬 글자 위치 --> ', str_pos_list)
print('프로그램3 종료')
```

```python
'''
한개의 조건문에 대해
True일 때 수행할 문장과 False일 때 수행할 문장이 다르다면

if 조건문:
    True일 때 수행
else:
    False일 때 수행할 문장
'''
```

```python
age = 15
if age <= 19:
    print('미성년자')
else:
    print('성인')

print('if문 종료')
```

```python
# 수행문이 한줄인 경우 : 옆에 작성 가능
x = 5
if x > 5 : print('5보다 큼')
else : print('5보다 크지 않음')
```

```python
# 비추천 / 한줄에 여러 문장을 적고 싶다 -> ; 으로 구분
if 10 > 5 : a = 1; print(a)
```

```python
# 임의의 정수를 num에 저장
# 4의 배수라면 '4의 배수입니다' 출력     / ~~라면 이라는 문장이 나오면 if라고 생각
# 아니라면 '4의 배수가 아닙니다.' 출력

num = int(input('정수를 입력하세요'))

if num % 4 == 0:
    print('4의 배수입니다.')

else:
    print('4의 배수가 아닙니다.')
```

```python
result = '4의 배수가 아님' # False인 경우 수행할 내용
num1 = int(input('정수 입력'))
if num1 % 4 == 0:
    result = '4의 배수' # True일 때만 바꿔준다
print(result)
```

```python
# pass : 특정 조건 시 아무 일도 하지 않고 넘어가야 할 때
x = 10
if x == 0:
    pass # 0인 경우 아무 일도 안하게 / 아무것도 안적으면 error가 나기 때문에 pass를 사용
else:
    print('값이 있습니다.')
```

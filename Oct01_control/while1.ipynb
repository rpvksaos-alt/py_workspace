```python
# while : 지정한 조건식 True인 동안 반복 실행
'''
while 조건식 : 
    수행문
    수행문...

'''
'''
while True:
    수행문
    if 조건:
        break
'''
```

```python
count = 1
while count <=5:
    print(f'현재 카운트: {count}')
    count += 1
else: 
    print('반복이 종료되고 난 다음 마지막 할 작업')
```

```python
# break : 반복 탈출
n = 1
while n <= 5: # 1,2,3,4,5
    print(n)
    if n == 2:
        break # n 값이 2와 일치하면 반복 중단
    n += 1
else: print('break가 돌면 else 수행 x')    

print('while문 끝남')    
```

```python
# 무한 루프와 break
# 의도적으로 무한 반복을 만들고 특정 조건에서 break로 탈출
# 사용자의 입력을 지속적으로 받아야 하는 경우 사용

while True:
    user_input = input('종료하려면 q를 입력하세요')

    if user_input == 'q':
        print('입력을 종료합니다.')
        break
    print (f'입력내용: {user_input}')    



print('end')
    
```

```python
answer = 7

# 사용자가 정답을 맞출때까지 숫자를 입력 받으세요
# 정답을 맞췄으면 정답입니다. 틀리면 다시 입력하세요 출력
print('숫자 맞추기 시작')
while True :
    num = int(input('정답'))

    if num == answer:
        print('정답입니다.')
        break
    else: # 굳이 else 입력 안하고 print만 써도 가능
        print('다시 입력하세요')

print('end')
                
```

```python
# 위와 같은 해결이지만 비추천 코드 / 참고용으로 볼 것

answer = 7
num = 0

print('숫자 맞추기를 시작합니다.')
while num != answer:
    num = int(input('숫자 입력:'))
    if num == answer:
        pass
    else:
        print(num, '다시 입력하세요')
else: print('정답입니다.')
```

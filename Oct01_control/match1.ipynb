```python
# 동일한 값으로 수행문을 선택하는 match
# 3.10 버전에 추가됨 (이전 버전은 사용x)

menu = input('메뉴 선택(1,2,3)')

match menu:
    case '1' : print('coffee')
    case '2' : print('water')
    case '3' : print('tea')
    case _: print('잘못된 메뉴입니다.')

# _ -> 어떤 값이든 해당
```

```python
# 오늘이 무슨 요일인지 출력
from datetime import datetime

today = datetime.today() # 현재 날짜와 시간

print(today.weekday()) # 요일 -> 0 ~ 6 (월요일이 0) 월0 화1 수2 목3 금4 토5 일6

w = today.weekday()
day = '일'
match w:
    case 0: day = '월'
    case 1: day = '화'
    case 2: day = '수'
    case 3: day = '목'
    case 4: day = '금'
    case 5: day = '토'
print(day)
# | : 여러 값중에 하나와 일치한다면 (or)
match w :
    case 0 | 1 | 2 | 3 | 4 :
        print('평일')
    case 5 | 6:
        print('주말')
```

```python
#2. 나이를 입력받아 요금을 출력하세요.
# 0 ~ 7세     5,000원
# 8 ~ 19세    8,000원
# 20 ~ 64세   12,000원
# 65세 이상    6,000원
```

```python
age = int(input('나이 입력'))
if age >= 0 and age <= 7:
    print('5,000원')
elif age >= 8 and age <= 19:
    print('8,000원')
elif age >= 20 and age <= 64:
    print('12,000원')

else : print('6,000원')
```

```python
#3. 세 개의 정수를 입력받아 가장 큰 숫자를 출력하세요.
```

```python
a = int(input('첫번째'))
b = int(input('두번째'))
c = int(input('세번째'))

max_value = a # 최대값 저장 공간

if max_value < b:
    max_value = b # 새로 비교하는 값이 더 크다면 새로운 값 저장

if max_value < c:
    max_value = c

print(f'{a},{b},{c} 중 가장 큰 숫자 : {max_value}')

# max() : 최대값을 찾아주는 파이썬 내장 함수
print(max(a,b,c))
# min () : 최소값을 찾아주는 파이썬 내장 함수
print('최소값: ',min(a,b,c))
```

```python
num1 = int(input('정수'))
num2 = int(input('정수2'))
num3 = int(input('정수3'))

if num1 > num2 and num1 > num3:
    print(num1)
elif num2 > num1 and num2 > num3:
    print(num2)
elif num3 > num1 and num3 > num2:
    print(num3)
```

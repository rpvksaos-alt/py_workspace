```python
# 조건(True/False)에 따라 코드를 수행하는 조건문(conditional Statement)
'''
if 조건식:
    조건식의 결과가 True일때 수행할 문장1
    조건식의 결과가 True일때 수행할 문장2
    조건식의 결과가 True일때 수행할 문장3
            

if문 상관 없이 실행 할 문장 - 들여쓰기를 끝내면 됨

* 코드블록의 범위를 결정 -> 들여쓰기(tab키)
'''
# :(콜론)을 찍고 다음 문장으로 넘어가면 자동으로 들여쓰기가 됨 안되면 tap키 (if 조건식 범위 정하기)
```

```python
if True:
    print('수행1')
    print('수행2')

print('if 상관없는 문장') 
```

```python
if 10: # bool타입의 값으로 자동 변환 / 0 이나 '' 이런 빈 문자열은 실행 안됨 
    print('수행')
```

```python
age = 10
if age <= 19:
    print('미성년자입니다.')
```

```python
# 논리 연산자 not and or (중요 암기)
'''
not True/False -> 의미를 반대로 바꿔줌. True -> False, False -> True
T/F and T/F : 모든 값이 True이어야 True. 모든 조건을 다 만족해야 하는 경우 사용
T/F or T/F : 주어진 값 중 하나라도 True면 True. 주어진 조건 중 하나 이상 만족하면 되는 경우 사용

not True -> False
not False -> True
not '' -> True 

F and F  F
F and T  F
T and F  F
T and T  T -> 모든 조건이 다 True 일때만 True
                       첫번째 값이 False 라면 뒤에 식은 실행하지 않는다
F or F  F
F or T  T  -> 하나라도 True가 있으면 True
T or F  T        첫번째 값이 True라면 뒤에 식은 실행하지 않는다
T or T  T
'''
```

```python
# 주말이거나 공휴일이면 쉰다
is_weekend = False
is_holiday = True

print(is_weekend or is_holiday)
```

```python
# 시험성적(score)이 80이상이고 출결(attendance) 점수가 70이상인 경우 통과

score = 85
attendance = 80

print( score >= 80 and attendance >= 70)
```

```python
# or : 첫번째 식이 True(값이 있다면)라면 뒤 연산 수행 안하고 해당 값 반환
# 첫번째 값이 없다면 다음 값으로 넘어감
# 기본값(값이 없을 때 대체하는) 설정용으로 많이 사용함
print('' or 'apple')
print('banana' or 'apple')
print(''or 0 or 'cherry')
```

```python
# input 창에 값을 입력하지 않고 엔터 누르면 ''이 전달된다.
# 이름을 입력하면 이름으로, 이름 입력 안했으면 게스트로 출력
name = input ('이름을 입력하세요') or '게스트'
print(f'{name}님 환영합니다.')
```

```python
# 연산자 : 정해진 타입의 값을 받아서 그 값을 가공해 반환
# 사칙연산자 + - * / // % **

num1 = 7
num2 = 2

print(num1 + num2)
print(num1 - num2)
print(num1 * num2)
print(num1 / num2)
print(num1 // num2) # 나누기의 몫(int)만 (나머지x)
print(num1 % num2) # 나머지(int)
print(num1 ** num2) # 제곱
```

```python
# 숫자형을 제외한 나머지 타입을 더하면 결합
s1 = 'hello'
s2 = 'world'
print(s1 + s2) # 결합 (concat)
print(s1 - s2) # +를 제외한 나머지 사칙 연산은 숫자형만 지원
```

```python
# str * int
print('-' * 50)
```

```python
# 나누기 연산 시 주의할 점
# 0, 0.0, false 으로 나눌 수 없음
 print(100 / False) # false -> 0

print(100 / 0) # //, % 마찬가지로 error
print(100 / 0.0)
print(100 // 0)
print(100 % 0)
```

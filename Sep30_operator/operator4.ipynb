```python
# in 연산자 : 시퀀스 자료형 안에 특정값이 포함되어 있는지 확인
# in : 값이 안에 포함되어 있는가
# not in : 값이 안에 포함되어 있지 않은가

# 시퀀스 자료형 : 문자열, list, tuple
#               데이터의 순서(sequence)가 있고 연속적으로 이어져 있는 구조의 데이터

text = "python is fun."
print('python' in text)
print('ZZZZ' in text)
```

```python
store = ['apple', 'banana', 'cherry', 'fig']
print('apple' in store)
print('berry' in store)
print('app' in store) # 리스트이 요소(element)와 일치 해야함
```

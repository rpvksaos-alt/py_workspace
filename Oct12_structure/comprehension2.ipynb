```python
# 필터링 (if)
# 변수  = [ 조건식이 True 일때만 수행문 for 변수 in 시퀀스 if 조건식]

# 0 ~ 9 사이의 짝수만 저장
even_list = [ i for i in range(10) if i % 2 == 0]
print(even_list)
```

```python
# 변수  = [True수행문 if 조건식 else False수행문 for 변수 in 시퀀스 ]
even2 = [ i if i % 2 == 0 else 'X' for i in range(10)]
print(even2)
```

```python
# 다중 for문
word1 = 'abcde'
word2 = '12345'

result = [ i+j for i in word1 for j in word2]
print(result)

'''

for i in word1:
    for j in word2:
        list.append(i+j)

'''
```

```python
result = [ i+j for i in word1 for j in word2 if '1' in i+j]
print(result)
```

```python
result = [ i+j if '3' in i+j else 'XXX' for i in word1 for j in word2]
print(result)
```

```python
# 이차원 리스트를 일차원 리스트로 변환(평탄화 - platten)
m = [[1,2],
     [3,4],
     [5]]
p = [ n for row in m for n in  row ]
print(p)
```

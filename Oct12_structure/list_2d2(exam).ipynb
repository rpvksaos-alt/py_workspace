```python
#모든 요소의 합계와 평균 구하기
scores = [
    [80, 90, 85],
    [70, 75, 80],
    [95, 90, 100]
]

total = 0
count = 0

for i in scores:
    for s in i:
        total += s
        count += 1
print(total)
print(total / count)
```

```python
# 행별 합계 구하기
scores = [
    [80, 90, 85],
    [70, 75, 80],
    [95, 90, 100]
]
index = 1
for score in scores:
    print(f'{index}번 행 : {sum(score)}')    
    index += 1
```

```python
# 다음 리스트에서 가장 큰 숫자를 찾아 출력하세요.
data = [[12, 45, 23], [67, 34, 89], [21, 56, 78, 5]]

max_value = data[0][0] # 비교값 선정
for i in range(len(data)): # 0,1,2
    for j in range(len(data[i])):
        if max_value < data[i][j]:
            max_value = data[i][j]

print('최대값 :', max_value)      
```

```python
#왼쪽 위부터 오른쪽 아래까지 대각선 요소의 합계를 구하세요.
matrix = [[1, 2, 3], 
          [4, 5, 6], 
          [7, 8, 9]]

total = 0
for i in range(len(matrix)):
    total += matrix[i][i]

print(total)    
```

```python
# 숫자 5가 몇 번 등장하는지 세어 보세요.
data = [[5, 2, 5], [1, 5, 3], [5, 4, 5]]

count = 0
for d in data:
    for num in d:
        if num == 5:
            count += 1

print(count)
```

```python
# 리스트의 요소를 가지고 올 때 인덱스도 같이 가지고 오자!!
# enumerate(리스트) : 순서대로 (인덱스, 값) 출력

list1 = ['A','B','C','D']
for i in enumerate(list1): # [(0,'A'), (1,'B'), (2,'C'), (3,'D')]
    #print(i)
    print(f'list1[{i[0]}] : {i[1]}')

print()

for i,v in enumerate(list1): # tuple 언패킹
    print(i,v)
```

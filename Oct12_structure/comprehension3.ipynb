```python
# 정수 리스트가 주어질 때, 짝수만 골라 제곱한 리스트
nums = [1, 2, 3, 4, 5, 6]

num1 = [ i ** 2 for i in nums if i % 2 == 0]
print(num1)
```

```python
# 문자열 리스트에서 문자열의 길이가 3이상인 문자열만 골라 저장한 리스트
words = ['hi','hello','a','안녕하세요']

w = [ i for i in words if len(i) >= 3]
print(w)
```

```python
# 정수 리스트에서 홀수는 그대로, 짝수는 'even'으로 변환된 리스트
nums = [1,2,3,4,5]

even = [ e if e % 2 !=0 else 'even' for e in nums]
print(even)
```

```python
#딕셔너리에서 값이 80이상인 학생의 이름만 리스트로 추출
scores = {'kim': 75, 'lee': 82, 'park': 90, 'choi': 68}

result4 = [ n for n,s in scores.items() if s >= 80]
print(result4)
```

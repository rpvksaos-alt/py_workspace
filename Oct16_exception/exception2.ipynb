```python
try:
    num = int(input('숫자 입력'))
    result  = 100 /num
except ValueError:
    print('숫자만 입력하세요')
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다.')
else:
    print('계산 결과:', result)
finally:
    print('프로그램 종료')
```

```python
try:
    for i in range(0,5):
        result  = 10 / i
        print(result)

finally:
    print('f') # 무조건 수행 후 강제 종료(except 없으니까)
```

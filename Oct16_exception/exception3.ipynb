```python
# 강제 예외 발생
# 이미 정해진 상황말고 내가 원하는 시점에 예외를 발생

# raise 예외 타입(예외 메세지)
# 해당 예외 메세지를 가지고 있는 예외 객체 발생 -> 예외 발생 -> 강제 종료
while True:
    value = input('변환할 정수값 입력')
    for d in value:
        if d not in '02468':
            raise ValueError('짝수 숫자가 아닌 값이 포함되어 있습니다. 다시 확인하세요')
    print(int(value))
```

```python
# assert 예외 조건
# 조건에 만족하지 않은 경우 예외 (AssertionError)발생
# assert False - 예외가 발생한다

# 정수 입력받아서 이진수 변환
# 정수가 아니라면 예외 발생
def get_binary_num(decimal_num):

    # isinstance(값, 검사할 타입) : 값이 해당 타입이면 True
    assert isinstance(decimal_num, int)
    return bin(decimal_num) # bin(십진수) : 이진수 변환

print(get_binary_num(10))
# 호출한 함수 안에서 예외 발생 -> 남은 문장 포기 -> 호출 지점으로 예외 객체 들고 와
# 함수를 호출하는 지점에서 예외처리 해준다
try:
    print(get_binary_num('10'))
except AssertionError:
    print('int타입의 값이 아님')
```

```python
# 양수 제곱 함수
def square_positive(n:int) -> int:
    #1. 숫자가 아닌 값 전달
    if not isinstance(n, int):
        raise TypeError('정수만 가능합니다.')
    #2. 0 
    assert n != 0, '0은 허용하지 않습니다.' # assert 조건, 조건 False일때 에러 메세지
    
    #3. 음수
    if n < 0:
        raise ValueError('양수만 입력 가능합니다.')

    return n ** 2

try:
    print(square_positive(5))
    print(square_positive(0))
except (TypeError, AssertionError, ValueError) as e:
# as e : except에서 처리한 에러 객체를 e라는 별칭으로 부른다.
    print('에러 발생:', e ) # 에러 객체 출력 시 에레 메세지 출력


print('프로그램 종료')
```

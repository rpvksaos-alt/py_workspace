```python
# 변수를 선언하고 값을 저장해보자
# 변수명 = 값 (초기화 해주기!!)
```

```python
x = 10 # int
name ='kim' # str
height = 170.5 # float
is_student = True #bool
```

```python
# 타입 확인하기
# type(변수 / 값)
print(f'{x}타입의 값은{type(x)}입니다.')
print(f'{name}타입의 값은{type(name)}입니다.')
print(f'{height}타입의 값은{type(height)}입니다.')
print(f'{is_student}타입의 값은{type(is_student)}입니다.')
```

```python
data = input('값을 입력하세요')
# input은 모든 값을 다 받아와야 하기 때문에 무조건 str 타입으로 받아온다.
print(data, type(data))
```

```python
# 동일한 값을 할당(독립된 변수)
x = y = z = 0
print(f'x:{x}, y:{y}, z:{z}')

z = 10
print(f'after z = 10 - > x:{x}, y:{y}, z:{z} ')
```

```python
# 다중 할당(여러개의 변수에 한번에 값 할당)
a, b, c = 1, 2, 3
print(f'a:{a}, b:{b}, c:{c}')
```

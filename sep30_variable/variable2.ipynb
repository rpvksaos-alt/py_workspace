```python
str1 = ('hello''world''python') # 하나의 문자열로 합쳐줌 ('문자열''문자열'...) , 없이 나열
print(str1)
```

```python
# 값 처리 원칙
# 같은 타입끼리 연산해야 함.
# 연산의 결과도 같은 타입이 나와야 함.
# 같은 타입끼리 대입해야 한다. (파이썬 해당 X)
```

```python
'''
타입 변환
- 연산자나 문맥에 따라 자동으로 변환
- 함수를 사용해 강제로 변환
 int(): (숫자로만 이루어진) 문자열 -> int, float -> int
 float(): (숫자와.으로만 이루어진)문자열 -> float
 str(): 문자열로 변환
 bool(): 값이 있다면 -> True, 값이 없다면 -> False
'''
```

```python
print(10 + 'abc') # TypeError - > str + int
```

```python
print(str(10) + 'abc') # 타입을 같게 변환해 + 처리함
# 문자열과 문자열을 더하면 이어쓰기가 됨
```

```python
#소수점 이하의 값을 버리고 싶을때
print(int(10.5))
print(int(3.1415))
```

```python
# 숫자형 간의 사칙연산
# int + float
# 데이터의 손실이 발생하면 안됨 -> int를 자동으로 float로 변환해서 연산
n1 = 10 # 10.0
n2 = 3.14
result = n1 + n2
print(result, type(result))
```

```python
# 나누기 연산자 때문에 (나머지가 있기 때문에) 결과물은 항상 float 타입이 나옴
n3 = 10
n4 = 2
result2 = n3 / n4
print(result2, type(result2))
```

```python
# 문자열(str)에 int를 곱하면 해당 문자열 반복 (사칙연산x)
print('ABC' * 3)
print('ABC' * int(3.5))
```

```python
# 숫자형으로 변환시 주의할 점
print('100 ->', int('100'))
# print('100.0 ->', int('100.0')) # 숫자로만 이루어져 있어야함
print('100.00 ->', float('100.00'))
#print('100.00 ->', float('100.0.0')) # 소수점도 한개 까지만 가능
#print('abc ->', int('abc'))
#print('True ->', int('True'))
print('True ->', int(True)) # 1
print('False ->', int(False)) #0
```

```python
# bool + bool -> 자동으로 int로 변환
print(True + True)
print(True + False)
print(True * True)
print(True - 10)
```

```python
#sum() : 전달 받은 값들을 다 합한 결과 반환
sum([True, True, False, False, True]) # - > True의 개수를 구할 수 있다.
```

```python
print(bool(10)) # True로 변환되는 값들(Truthy)
print(bool(-10))
print(bool(0)) # False로 변환되는 값들(Falsy)
print(bool(0.0))
print(bool('')) # 빈 문자열
print(bool(None))  # None : 값이 없음. 존재하지 않음
print(bool([])) # 빈 리스트
```

```python
# 학생 이름과 점수를 입력받아 students.txt에 저장하고, 
# 저장된 내용을 다시 읽어 출력
'''
[students.txt]
김스캇,85
이제임스,92
'''
```

```python
# 선생님 답

def insert_student():
    # 학생 정보 입력
    name = input('학생이름')
    score = int(input('학생 성적'))
    # 파일에 저장
    with open('students.txt', 'a', encoding='utf-8') as file:
        file.write(f'{name},{score}\n')
    print('저장 완료')

# 파일 읽기
def read_students():
    with open('students.txt', 'r', encoding='utf-8') as file:
        students = file.readlines()
        for s in students:
            name, score = s.split(',')
            print(f'{name},{score}')

insert_student()
insert_student()
read_students()
```

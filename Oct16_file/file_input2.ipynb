```python
with open('dream.txt') as file:
    contents = file.read()

print(contents) 
```

```python
with open('dream.txt','r') as file:
    contents = file.readlines()

print(contents[:2])
print(f'dream.txt는 총 {len(contents)}줄 입니다.')
```

```python
with open ('dream.txt','r') as file:
    i = 1
    while True:
        line = file.readline().replace('\n','')
        if not line: # 더이상 읽어온 내용이 없으면 종료
            break
        if line.strip() == '':# 읽어온 줄이 빈줄이라면 # strip : 공백제거
            continue
        print(f'{i} === {line}')
        i += 1
```

```python
# 파일의 글자 통계 정보 출력하기
# 전체 글자 수, 단어 수, 줄 수
with open('dream.txt','r') as file:
    contents = file.read()
    word_list = contents.split(' ')
    line_list = contents.split('\n')
    

print('글자 수',len(contents))
print('단어 수',len(word_list))
print('줄 수',len(line_list))
    
```

```python
# utf-8로 인코딩 된 파일을 cp949방식으로 디코딩 -> error

# UnicodeDecodeError:
# 'cp949' codec can't decode byte 0xec in position 0: 
# illegal multibyte sequence
with open('문서1.txt','r') as f:
    contents = f.read()
    print(contents)
```

```python
# 인코딩 : 문자 -> 0,1 바이트로 바꿔서 저장
# 디코딩 : 바이트 -> 문자

# character-set
# 아스키코드 : 숫자, 특수문자, 영어 대소문자를 몇 바이트로 저장할 것인가의 약속
# cp949 : 한국어 window 환경에서 사용하는 인코딩 방식
# utf-8 : 한글뿐만 아니라 다국적 언어를 몇 바이트로 표기할 것인가 인코딩 방식

# 인코딩 지정 - 한글 문서는 encoding='utf-8' !!
with open('문서1.txt', 'r', encoding='utf-8') as f:
    text = f.read()
    print(text)
```

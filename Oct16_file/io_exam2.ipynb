```python
# 학생 이름과 점수를 입력받아 students.txt에 저장하고, 
# 저장된 내용을 다시 읽어 출력
'''
[students.txt]
김스캇,85
이제임스,92
'''
def insert_student():
    # 학생 정보 입력
    name = input('학생이름')
    score = int(input('학생 성적'))
    # 파일에 저장
    with open('students2.txt', 'a', encoding='utf-8') as file:
        file.write(f'{name},{score}\n')
    print('저장 완료')

# 파일 읽기
def read_students():
    with open('students2.txt', 'r', encoding='utf-8') as file:
        students = file.readlines()
        for s in students:
            name, score = s.split(',')
            print(f'{name},{score}')

insert_student()
insert_student()
read_students()
```

```python
# # 메모장 만들기
# 1. 사용자로부터 문장을 입력받아 memo.txt에 저장
# 기존 파일이 있다면 내용 덮어쓰기
memo = input('메모 입력')
with open('memo2.txt', 'w', encoding='utf-8') as file:
    file.write(memo)
print('메모 저장 완료')
```

```python
# 2. 메모 추가하기
# 사용자로부터 새로운 문장을 입력받아 memo.txt 마지막에 추가
memo = input('메모 입력')
with open('memo2.txt', 'a', encoding='utf-8') as file:
    file.write('\n'+memo) # 줄 바꾸고 추가
print('메모 추가 완료')

```

```python
# 3. 파일 내용 출력하기
# memo.txt파일 내용 전부 출력하기

with open('memo2.txt', 'r', encoding='utf-8') as file:
    content = file.read()
print(content)
```

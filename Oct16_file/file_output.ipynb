```python
# 파일 출력
# 파일로 내용을 내보낸다(write)

# open
# write() : str 한개를 파일로 출력.
# writelines() : list, tuple을 받아서 한번에 출력
# close | with
```

```python
# open(r'경로','모드',encoding='utf-8')
''' 
모드
'r' : 읽기 모드 (기본값)
'w' : 쓰기(기존 파일 내용 사라짐)
'a' : 이어쓰기 (기존 내용 뒤에 추가)
'r+' : 읽기/쓰기 겸용
't' : 텍스트 모드
'b' : 이진 모드(이진 파일 처리)
'''

# 읽기 -> 파일이 존재해야 함. 없으면 에러
# 쓰기 -> 파일이 없으면 새로 만들어서 쓰기. 있으면 있는 파일 활용
```

```python
f = open('count.txt','w', encoding='utf-8')
for i in range(20,31):
    data = f'{i}번째 줄 입니다.\n'
    f.write(data)
f.close()
```

```python
f = open('count.txt','a', encoding='utf-8')
for i in range(31,41):
    data = f'{i}번째 줄 입니다.\n'
    f.write(data)
f.close()
```

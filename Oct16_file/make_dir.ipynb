```python
# file : 컴퓨터를 실행할 때 가장 기본이 되는 단위
#        실제 데이터가 담긴 곳(데이터의 집합)

# directory : 파일이 모여 있는 곳
#            파일과 하위 디렉토리를 담아서 관리하는 공간
```

```python
# 폴더 만들기
import os
print(os.listdir()) # 현재 위치의 하위 폴더명과 파일명 목록
```

```python
# 폴더 존재 여부 판단
print(os.path.isdir('log'))

if not os.path.isdir('log'):
    os.mkdir('log') # 폴더 생성
else: print('log 폴더가 이미 있음')    
```

```python
# 파일 존재 여부 판단
# os.path.exists('경로')
print(os.path.exists('dream.txt'))
print(os.path.exists('log/count.txt'))
```

```python
# 파일이 없으면 새로 만들어서 시작 ('w')
# 파일이 있으면 기존 내용 뒤에 추가 ('a')
if not os.path.exists('log/count.txt'):
    f = open('log/count.txt','w', encoding='utf-8')
    f.write('log 메세지 시작')
    f.close()
else:     
    f = open('log/count.txt','a', encoding='utf-8')
    f.write('계속 log 출력1\n')
    f.write('계속 log 출력2')
    f.close()
```

```python
# os.remove('파일경로')

# FileNotFoundError : 지정한 경로에 파일이 없는 경우
# PermissionError : 파일 삭제 권한이 없는 경우

# 1. if문 사용해서
if os.path.exists('test.txt'):
    os.remove('test.txt')
    print('삭제 완료')
else: print('파일이 존재하지 않습니다.')

# 2. try 사용해서
try:
    os.remove('test.txt')
    print('삭제 완료')
except FileNotFoundError:
    print('없는 파일입니다.')
except PermissionError:
    print('파일 삭제 권한이 없습니다.')
```

```python
# 함수
# 두개의 숫자 값을 받아 더한 결과를 반환하는 add함수
def add(n1:int, n2): # :int는 숫자값을 받아온다는 힌트 (없어도 됨)
    result = n1 + n2
    return result
    n1 = 10 # return 아래 문장 수행 x
    print(n1)
```

```python
# 함수 실행
r = add(10, 20)
print(r)
```

```python
# 매개변수와 값이 맞이 않을 때 -> error
r = add(10, 20, 30)
```

```python
plus = add # 함수 객체 자체를 저장
print(plus)
print(plus(1,5)) # 함수를 저장한 변수명으로 함수 호출 가능
```

```python
# 동일한 이름의 함수가 또 선언됐다면?
# 3개의 값을 받아 다 더한 결과를 반환하는 add함수
def add(n1, n2, n3):
    return n1 + n2 + n3
```

```python
# 동일한 이름의 함수가 여러개 있다면
# 제일 마지막에 선언한 함수만 메모리에 남아 있음

# r = add(10,20) # 매개변수 2개짜리 add 사라짐
r = add(10, 20, 30)
print(r)
```

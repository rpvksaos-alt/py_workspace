```python
#2. 이름을 매개변수로 전달받아 "이름님, 환영합니다!" 형식으로 출력하는 greet(name) 함수

def greet(name):
    print( f'{name}님 환영합니다!')

greet("김종현")
```

```python
#4. 정수 하나를 전달받아 그 수의 제곱을 출력하는 square(n) 함수
def square(n):
    print(n ** 2)
square(5)
```

```python
#5. 정수 하나를 전달받아 짝수이면 "짝수", 홀수이면 "홀수"를 출력하는 check_even(n) 함수
def check_even(n):
    if n % 2 == 0:
        print('짝수')
        
    else:
        print('홀수') 
check_even(3)
```

```python
#6. 두 정수를 전달받아 더 큰 수를 return으로 반환하는 get_max(a, b) 함수

def get_max(a,b):
    if a > b:
        return a
    else:
        return b
print(get_max(5,6))       
```

```python
#7. 세 개의 점수를 전달받아 평균을 반환하는 get_average(a, b, c) 함수

def get_average(a, b, c):
    return (a + b + c) / 3

print(int(get_average(50,60,40)))
```

```python
#8. 상품 가격과 할인율을 전달받아 할인된 가격을 반환하는 get_discount(price, rate) 함수
#(할인율은 백분율로 전달. 예 20% 할인은 20)
def get_discount(price, rate):
    return price - (price * rate / 100)

print(int(get_discount(10000, 10)))
```

```python
#10. 두 숫자와 연산자(+, -, *, /)를 전달받아 계산 결과를 반환하는 calculator(a, b, op) 함수
# 나눗셈에서 두 번째 숫자가 0이면 "0으로 나눌 수 없습니다."를 출력

def calculator(a, b, op):
    if op == '+':
        return a + b
    elif op == '-':
        return a - b
    elif op == '*':
        return a * b
    elif op == '/':
        if b == 0:
            print('0으로 나눌 수 없습니다.')
            return
        return a / b
    else :
        print('잘못된 연산자 입니다.')
        return

# 호출 코드
# print(calculator(10, 5, "+"))
# 15
print(calculator(10, 5, "+"))
print(calculator(10, 5, "-"))
print(calculator(10, 5, "*"))
print(calculator(10, 5, "/")) 
print(calculator(10, 5, "/////"))
print(calculator(10, 0, "/"))
```

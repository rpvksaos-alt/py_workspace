```python
# 7.특정 문자열 포함 여부
# 다음 문자열에 "admin"이 포함되어 있는지 확인하고
# 포함되어 있으면 "관리자", 아니면 "일반 사용자"를 출력하세요.


role = "superadmin"

# 1. 함수 사용
if role.find('admin') >= 0:
    print('관리자')
else:
    print('일반 사용자')

# 2. in 연산자 사용
if 'admin' in role:
    print('관리자')
else:
    print('일반 사용자')
```

```python
# 8.숫자로 이루어진 문자열 처리
# 입력값을 정수로 변환하여 100을 더한 결과를 출력하세요.
# int() 사용이 가능한 값인지 조건 검사하세요
# 조건 통과 못하면 '잘못된 값입니다.'



num:str = input('숫자 입력 > ')
if num.isdigit(): # 음수, 실수 X
    print(int(num)+ 100)
else: print('잘못된 값 입니다.')

# 음수 처리
flag = True # True :int 변환시 문제 없음, False : 문제 있음
for n in num:
    if n not in '0123456789-':
        print('잘못된 값 입니다.')
        flag = False
        break
if flag:    
    print(int(num) + 100)

for n in num:
    if n not in '0123456789-':
        print('잘못된 값 입니다.')
        flag = False
        break
    else:
        print(int(num) + 100)
```

```python
# 9.비밀번호 검증
# 다음 조건을 만족하면 "통과", 아니면 "실패"를 출력하세요.
# [조건]
# 1.길이 8 이상
# 2."!" 문자를 포함

password = 'password!'

if len(password) >= 8 and '!' in password:
    print('통과')
else:
    print('실패')

```

```python
#문자열이 주어졌을 때 모음(a, e, i, o, u)의 개수를 출력하세요.
text = "Python Programming"

count = 0

for ch in text.lower():
    if ch in "aeiou":
        count += 1

print(count,'개')
```

```python
# 문자열이 주어졌을 때 대문자는 소문자로, 소문자는 대문자로 바꾸세요.
text = "Hello Python"

result = '' # 빈문자열로 초기화

for ch in text:
    if ch.isupper():
        result += ch.lower()
    else:
        result += ch.upper()
print(result)

#result = text.swapcase()
#print(result)
```

```python
# 문자열이 회문(palindrome)인지 판별하세요.
# 앞에서 읽어도, 뒤에서 읽어도 같은 문자열이면 YES, 아니면 NO를 출력합니다.

text = "level"


r_text = text[::-1] # 역순

if text == r_text:
    print('yes')
else:
    print('no')
```

```python
# 문자열에서 숫자만 추출해서 하나의 문자열로 만드세요.
text = "abc123def45"
result = ""
for ch in text:
    if ch.isdigit():
        result += ch
print(result)
```

```python
# 문자열에서 중복된 문자를 제거하고 처음 등장한 순서를 유지하세요.
text = "banana"

result = "" # 없으면 추가

for ch in text:
    if ch not in result:
        result += ch
print(result)

#result = "".join(dict.fromkeys(text))
#print(result)
```

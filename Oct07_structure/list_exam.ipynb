```python
#1. 다음 리스트에서 짝수만 출력하세요.
numbers = [12, 7, 25, 8, 14, 31, 20, 9]

for i in numbers:
    if i % 2 == 0:
        print(i, end=' ')
        

```

```python
#2. 다음 리스트에서 홀수만 새로운 리스트에 저장하세요.
numbers = [3, 8, 15, 20, 7, 12, 9]

result = [] # 새로운 리스트

for n in numbers:
    if n % 2 != 0: # 홀수일 때
        result.append(n) # 리스트에 추가
print(result)
```

```python
#3. 다음 리스트에서 10보다 큰 숫자의 개수를 구하세요.
numbers = [5, 12, 7, 20, 3, 18, 9, 25]

count = 0 # 개수

for n in numbers:
    if n > 10:
        count = count + 1 # count += 1
print(count)
```

```python
#4. 양수와 음수의 개수를 각각 구하세요.
numbers = [10, -3, 7, -8, 0, 15, -2, 4]

p_count = 0
n_count = 0

for n in numbers:
    if n > 0:
        p_count += 1
    elif n < 0:
        n_count += 1
print(f'양수 개수 : {p_count}개, 음수 개수 : {n_count}개')
```

```python
#5. 가장 큰 숫자가 몇 번째 인덱스에 있는지 출력하세요.
numbers = [10, 25, 7, 42, 18, 30]

max_n = numbers[0] # 첫번째 숫자부터 /[0] <-첫번째 자리
max_index = 0 # 첫번째 숫자의 인덱스
for i in range(1, len(numbers)): # 1번 인덱스부터 마지막 인덱스까지
    if numbers[i] > max_n:
        max_n = numbers[i]
        max_index = i
print(max_n)       
print(max_index)
```

```python
#6. 중복을 제거하세요. 처음 등장한 순서를 유지해야 합니다.
numbers = [1, 3, 3, 5, 1, 7, 5, 9, 3]

result = [] # 중복 없는 값 저장할 공간(순서 유지 -> 리스트)

for n in numbers:
    if n not in result:
        result.append(n)
print(result)        
```

```python
#7. 회원 이름 목록중 '김'으로 시작하는 회원만 따로 저장하세요
users = ["김민수", "이영희", "김철수", "박지민", "김수진"]

result = []
for name in users:
    # if name[0] == '김':
    if name.startswith('김'):
        result.append(name)

print(result)        
```

```python
#8. 다음은 로그인 시도 횟수입니다.
login_count = [1, 3, 2, 10, 1, 7, 2, 15, 4]
#로그인 시도가 5회 이상인 사용자 수를 구하세요

count = 0

for login in login_count:
    if login >= 5:
        count += 1
print('로그인 5회 이상 시도한 유저 수:',count) 
```

```python
#9. 오늘의 판매 데이터 분석

sales = [12000, 35000, 12000, 50000, 18000,
         35000, 70000, 12000, 25000]

# 1. 총 판매 금액
total_price = 0
for money in sales:
    total_price += money
print(total_price)
print(sum(sales)) # sum(리스트) : 리스트 내의 값의 합계

# 2. 판매 건수
count = len(sales)
print(count)
# 3. 평균 판매 금액
average = total_price / count
print(int(average))
# 4. 30000원 이상 판매 건수
over_count = 0
for money in sales:
    if money >= 30000:
        over_count += 1
print(over_count)        
        
```

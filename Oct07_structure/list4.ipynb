```python
# 리스트에서 자주 사용하는 기능(함수)
```

```python
# 특정 값의 개수 구하기
a = [1,2,4,6,0,2,5,6,7]
b = a.count(6) # 2
print(b)

b = a.count(10) # 없으면 0
print(b)
```

```python
# 순서 뒤집기
print(a[::-1])
print(a) # 원본 데이터 유지 -슬라이싱 처리 저장 x
#역순 저장
b = a[::-1] # 변수에 저장
print(b)

a.reverse() # 원본 리스트의 순서를 바꿔버림
print(a)
```

```python
# 정렬
# sort() : 원본 데이터 정렬(오름차순/내림차순)
c = [10,20,50,60,30,40]
c.sort() # 오름차순 정렬
print(c)

c.sort(reverse=True) # 내림차순 정렬
print(c)
```

```python
# 원본데이터를 남기고 정렬된 새로운 리스트 반환
c2 = [10,20,50,60,30,40]

c3 = sorted(c2)
print(c2) # 원본 리스트 그대로
print(c3) # 새로운 오름차순 정렬된 리스트

c4 = sorted(c2, reverse=True) # 새로운 내림차순 정렬된 리스트
print(c4)
```

```python
# 모든 요소 삭제하기
d = [1,2,3,4]
d.clear()
print(d) # 빈 리스트
```

```python
# 리스트 복사
a = [0,0,0,0,0]
b = a # 주소값 복사
      # 하나의 리스트를 같이 사용  

print(a)
print(b)

b[2] = 99

print(a)
print(b)
```

```python
# copy()
# hard copy : 똑같은 데이터를 가지고 있는 리스트를 하나 더 생성

a2 = [0,0,0,0,0]
b2 = a2.copy() # 주소는 다르고 같은 값을 가지고 있는 리스트 생성

print(a2)
print(b2)

print(a2 == b2) # 리스트 내 내용 비교
print(a2 is b2) # 리스트의 주소값 비교

b2[2] = 99
print(a2)
print(b2) # b2가 가리키고 있는 리스트만 변경
```

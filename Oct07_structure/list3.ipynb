```python
# list 함수 (기능)

list1 = ['AAA','B','C','D','E']
# len(리스트) : 리스트에 저장된 element의 수를 반환
print(len(list1))
print('list1의 마지막 인덱스 번호:', len(list1)-1) #0,1,2,3,4
```

```python
# 리스트 내장함수 -> 리스트.함수()

# 리스트에 데이터 추가
# 리스트.append(값) : 기존 리스트 맨 끝에 추가
# 리스트.insert(인덱스, 값) : 인덱스 지점에 값 추가. 기존 데이터는 1자리씩 밀림
# 리스트.extend([리스트]) : 기존 리스트 안에 전달받은 리스트의 값들을 추가
```

```python
print('list1:', list1)

list1.append('zz') # 마지막 자리에 추가
print('append("zz")', list1)

list1.insert(2,'yyyy') # 원하는 자리에 추가
print(list1)

list1.extend(['apple','banana']) # 한번에 여러개(리스트) 추가
print(list1)
```

```python
list2 = []
list2.append('A1')
list2.append('A2')
list2.append('A3')
print(list2)
```

```python
# 리스트 삭제
''' 
리스트.remove(값) : 리스트 내 특정 값 삭제. 없는 데이터 삭제 시 에러
                   중복값인 경우 제일 앞 데이터 삭제
변수 = 리스트.pop() : 제일 마지막 데이터를 꺼내오면서 list에서는 삭제                   
리스트.pop(인덱스) : 인덱스에 있는 값 꺼내오면서 삭제
del 리스트[인덱스] : 특정 인덱스의 값 삭제

값이 삭제되면 뒤에 있던 데이터가 앞으로 당겨짐
'''
```

```python
list3 = [10, 11, 12, 10]
list3. remove(10) # 동일한 값을 삭제
print(list3) # 뒤에 있는 10은 삭제되지 않음

list3.remove(100) # ValueError: list.remove(x): x not in list
```

```python
list5 = list(range(5)) # [0,1,2,3,4]
del_data = list5.pop() # 인덱스 지정 안하면 -1 인덱스(마지막 데이터) 삭제
print(del_data, '삭제 후 ->', list5) # [0,1,2,3]
del_data2 = list5.pop(0) # 지정한 인덱스에 있는 값 꺼내면서 삭제
print(del_data2, '삭제 후 ->', list5) # [1,2,3]
```

```python
list6 = [1,2,3,4,5]
del list6[0]
print('0번 인덱스 삭제', list6)

# 부분삭제 - 슬라이싱
del list6[1:3]
print('[1:3]삭제', list6)

del list6[:] # 리스트 내 데이터 전부 삭제
print(list6)
```

```python
# 또 다른 삭제 방법
list7 = [1,2,3,4,5]
list7[2:4] = [] # 2 ~ 4번 앞까지 삭제
print(list7)
```

```python
''' 
del 리스트[인덱스] : 리스트 내 해당 인덱스에 위치한 값 삭제
del 변수명 : 메모리에서 해당 변수 삭제(더이상 사용x)
'''

a = 10
print(a)

del a
# print(a) # NameError: name 'a' is not defined

a = 20
print(a)
```

```python
# 리스트 수정(변경)
# 리스트[인덱스] = 값
list8 = ['apple', 'banana','cherry','melon']
list8[2] = '수박'
print(list8)

#list8[3:5] = 100 # 슬라이싱으로 변경하는 경우 list 타입 사용
list8[3:5] = [100]
print(list8)

list8[2:3] = [100, 200, 300]
print(list8)
```

```python
''' 
맨 뒤에 그냥 붙이고 싶다. append
원하는 위치에 끼워넣고 싶다. insert
삭제된 값을 변수에 받아서 쓰고 싶다. pop
값 자체를 지정해서 지우고 싶다. remove
슬라이싱 범위로 지우거나 변수 자체를 없애고 싶다. del
'''
```

```python
# 1. 선언 (declaration)
# '사과','바나나','우유' 세개의 문자열 요소를 가진 cart 리스트를 선언

# 2. 추가(addition)
# cart 리스트에 맨 뒤에 '계란'을 추가하세요

# 3. 수정(modification) 
# 인덱스 1번에 있는 '바나나'를 '포도'로 수정하세요

# 4. 삭제(Deletion)
# cart에서 '우유'를 삭제하세요

# 최종 결과 ['사과', '포도', '계란'] 출력
```

```python
# 1
cart = ['사과','바나나','우유']
print(cart)

# 2
cart.append('계란')
print(cart)

# 3
cart[1] = '포도'
print(cart)

# 4
cart.remove('우유')
print(cart)
```

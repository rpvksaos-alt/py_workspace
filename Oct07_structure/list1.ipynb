```python
'''
list : 순서가 있는 데이터의 집합(index), 중복 허용, 변경 가능
       인덱스로 원하는 데이터에 접근
'''
```

```python
# 선언(생성)
# 변수 = [값1, 값2, 값3 ...]
# 주소값들을 모아서 저장하고 있는 형태

list1 = [10, 20, 30, 40]

# 인덱싱 리스트[인덱스]
print(list1[0])

print(list1, type(list1))
```

```python
list2 = ['abc',123, 3.14, True] # 서로 다른 타입도 묶어줄 수 있음
list3 = [[10,20],[30,40]] # 리스트 안에 리스트 선언도 가능(이차원 리스트)
# 1차원 리스트이 주소값들을 모아서 가지고 있는 형태

print(list3[1][1]) # 1차원 리스트를 먼저 찾고 1차원 리스트에서 인덱싱
print(list3[0])
```

```python
# list() : 다른 타입의 자료구조를 받아와 list로 변환하는 용도로 주로 사용
list4 = list('hello')
print(list4)
```

```python
list5 = list(range(10)) # range() 사용해서 리스트 생성 가능
print(list5)
```

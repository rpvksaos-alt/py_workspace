```python
# 매개변수(parameter) : 전달받은 값 저장 공간
# 인자(argument) : 작업할 때 전달 받는 값
```

```python
# 키워드 인자
# 함수 호출 시에 argument 전달 -> 저장할 매개변수 이름을 지정해서 전달
# 이름(name)과 직책(position)을 전달받아 인사말을 출력하는 greet 함수

def greet(name, position):
    print(f'{name} {position}님 안녕하세요')

greet('김종현', '부장')
greet(position='팀장', name='김종현') # 매개변수 이름에 맞춰서 저장
# print('출력 내용', end = '', sep='/')

# 주의할 점 - 키워드 인자는 뒤에 명시
# greet(name = '김제임스', '프로') # 명시하지 않은 값은 순서대로 들어감 
    # name에는 '김제임스' 저장, '프로'는 지정하지 않았으니 순서대로 저장되는데 이미 name는 값이 있기 때문에 error

# greet('김제임스', name ='프로') # 키워드가 없는 값은 무조건 순서대로 들어감
                                # name에 값이 여러개 들어가는 상황 -> error

greet('김제임스', position ='프로')
```

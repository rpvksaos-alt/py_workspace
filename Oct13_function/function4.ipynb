```python
# * 값,값,값 -> 패킹
# * 자료구조 -> 언패킹

# 가변 인자 : 값이 몇개 들어올지 모르는 경우
#            여러개의 값을 받아 패킹해서 저장
#            제일 마지막 매개변수 앞에 *를 붙여줌 (한번만 선언 가능)
def astrisk_test(a, b, *args): # ()로 패킹해서 저장
    print(f'{a}, {b}, {args}')

astrisk_test(1,2,3)
astrisk_test(1,2,3,4,5,6)
astrisk_test(1,2)
# astrisk_test(1) # b에 저장할 값이 없음 -> error
```

```python
def astrisk_test2(*args):
    x,y,*z = args
    print(x,y,z)

astrisk_test2(3,4,5,10,20)
```

```python
# *values -> 여러개를 한번에 입력할 수 있음
print(1,2,3,4)
```

```python
# 기본값이 있는 매개변수와 같이 선언한다면
# *가 앞에 와도 됨
def astrisk_test3(*args, a=10, b=20):
    print(args)
    print(a,b)

astrisk_test3(1,2,3)
astrisk_test3(1,2,3, a = 50) # a, b에 저장하고 싶으면 키워드를 꼭 붙여야 함
astrisk_test3(1,2,3,50)
```

```python
# 키워드 가변 인수 - 이름이 있는 입력값들을 모아서 딕셔너리로 패킹해서 저장
def two_asterisk_test(**kargs):
    print(kargs)

two_asterisk_test(First = 1, Second = 2, Third = 3)
```

```python
def dict_funs(**params):
    for k in params.keys():
        print(f'{k} --> {params[k]}명 입니다.')

dict_funs(블랙핑크 = 4, 아이브 = 6, 에스파 = 4, NMIXX = 6)
```

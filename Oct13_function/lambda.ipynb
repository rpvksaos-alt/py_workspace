```python
# 람다함수 : 익명 함수(함수 이름 없음)
#           - 변수에 저장해서 변수명으로 호출
#           - 선언과 동시에 사용(선언된 위치에서 한번 사용하고 끝)

# 일반함수
def f(x,y):
    return x + y
f(1,2) # 이름으로 호출

# 람다함수
# 변수 = lambda 매개변수 : 수행문
f1 = lambda x, y : x + y # return 생략
print(f1(3,4))

f2 = lambda : 10
print(f2())

f3 = lambda a : a + 10
print(f3(10))
```

```python
# (람다함수 선언)(호출)
print((lambda x : x + 1)(5))
```

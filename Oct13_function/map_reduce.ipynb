```python
# 맵 리듀스
''' 
빅데이터를 처리하는 기본 알고리즘

맵(map) : 연속 데이터를 저장하는 시퀀스 형에서
모든 요소(element)마다 똑같은 작업을 하는 것

'''

# map() 함수
ex = [1,2,3,4,5]
# list(map(함수명, 리스트))
# 리스트에서 하나씩 뽑아서
# 함수에 전달
# 리턴 값을 모아서 list로 생성
def squere(x):
    return x ** 2

print(list(map(squere, ex)))
print(list(map(lambda x : x * x, [1,2,3])))

# 한줄로 모든 요소를 세제곱한 리스트
print([ i ** 3 for i in ex])
```

```python
# filter(조건 함수, 시퀀스 객체)
# 조건 함수 - 값을 하나씩 받아서 조건 검사 - T/F 반환
# 함수의 결과가 True 인 요소만 남김
numbers = [1,2,3,4,5,6]
def is_even(x):
    return x % 2 == 0 # 짝수라면 True, 아니면 False

result = filter(is_even, numbers)
print(list(result))
```

```python
employees = [
    {"name": "김철수", "department": "개발팀"},
    {"name": "이영희", "department": "인사팀"},
    {"name": "박민수", "department": "개발팀"}
]

# filter()를 사용해 개발팀 직원의 이름을 출력하세요
# 개발팀 직원만 필터링
def emp(e):
    return e['department'] == '개발팀'
result1 = filter(emp, employees)

for f in result1:
    print(f['name'])
```

```python
''' 
리듀스(reduce) : 요소를 하나씩 꺼내와 연산(함수) 후 결과를 누적하여 하나의 값으로 만들어내는 것
이전 연산 결과를 다음 연산으로 넘겨서 계속 연산이 이루어짐
'''

from functools import reduce # functools 모듈의 reduce 함수 사용하겠다

# result = reduce(함수, 리스트)
# 함수는 반드시 매개변수 2개 - 첫번째 매개변수:이전 연산 결과, 두번째 매개변수 : 새로운 element


# 1 ~ 5 까지의 합
ex = [1,2,3,4,5]
add_list = reduce(lambda x,y : x+y, ex)
print(add_list)

# 1 ~ 5 까지의 곱
def m(a,b):
    return a * b
print(reduce(m,ex))

multiply_list = reduce(lambda x,y : x*y, ex)
print(multiply_list)

# 최소값 구하기
# 1
nums1 = [45, 12, 89, 33, 67]
num_min = reduce(lambda x,y: x if x < y else y, nums1)
print(num_min)
# 2
min2 = reduce(min,nums1)
print(min2)
# 3
print(min(nums1))

# 문자열 합치기
words = ['python', 'is', 'awesome']
words_list = reduce(lambda x,y : x+' '+y, words)
print(words_list)

sentence1 = ' '.join(words)
print(sentence1)
```

```python
''' 
class 명 : BankAccount(은행계좌)
소유주(owner), 잔액(balance)
입금(deposit) : 입금액을 받아와 잔액에 더하고, '입금완료' 출력
출금(withdraw) : 출금액을 받아와 잔액에서 빼준다.
                 이때 잔액보다 출금액이 더 크면 빼지 않고 '출금 불가' 출력
                 아니면 출금 후 '출금 완료' 출력
잔액조회(get_balance) : 현 계좌의 잔액 반환
계좌정보 출력(__str__) : '예금주: {이름}, 잔액: {잔액}원' 출력
'''
''' 
1. 'kim'의 10만원 저금된 계좌 생성
2. 만원 입금
3. 잔액 조회
4. 출금 만원
5. 출금 20만원
6. 계좌 정보 확인
'''
```

```python
class BankAccount:
    def __init__(self,o):
        self.owner = o
        self.balance = 0

    def deposit(self, s):
        self.balance += s
        print('입금완료')
    def withdraw(self, s):
        if s > self.balance:
            print('출금불가')
        else:            
            self.balance -= s
            print('출금완료')
        
    def get_balance(self):
        return self.balance

    def __str__(self):
        return f'예금주{self.owner}, 잔액{self.balance}원'

w = BankAccount('kim')
print(w)
w.deposit(10000)
print(w)
w.withdraw(10000)
print(w)
w.withdraw(10000)
print(w)

print(w.get_balance())
                
```

```python
# 선생님 답
class BankAccount:
    def __init__(self, o:str, b:int):
        self.owner = o
        self.balance = b
    def deposit(self, money):
        self.balance += money
        print('입금 완료')
    def withdraw(self, money):
        if money > self.balance:
            print('출금 불가')
        else:
            self.balance -= money
            print('출금 완료')    

    def get_balance(self):
        return self.balance

    def __str__(self):
        return f'예금주: {self.owner}, 잔액:{self.balance}원'


kim_account = BankAccount('kim', 100000)
kim_account.deposit(10000)
print('현재 잔액:',kim_account.get_balance(), '원')
kim_account.withdraw(10000)
print('현재 잔액:',kim_account.get_balance(), '원')
kim_account.withdraw(200000)
print('현재 잔액:',kim_account.get_balance(), '원')
print(kim_account)    
```

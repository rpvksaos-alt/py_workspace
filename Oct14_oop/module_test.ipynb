```python
# 모듈 import 시
# Lib 폴더 -> 경로 상관없이 바로 가져다 사용

# 실행 파일 위치를 기준으로
# 같은 경로 -> 파일 이름부터 시작
# 하위 경로 -> 폴더명부터 시작

# 모듈로 사용할 패키지는 최상위 폴더에 만들자..
# 경로 찾기 쉽도록..
```

```python
import sys
from pathlib import Path

# Path.cwd() : 현재 실행 위치
# ipynb - 노트북 파일 : 실행 위치 -> 파일 위치
# py - 터미널에서 동작 -> 터미널의 위치

# 현재 실행 파일 폴더의 상위 폴더
workspace = Path.cwd().parent
# Path.cwd () -> C:\Users\soldesk\py_workspace\Oct14_oop
# parent -> C:\Users\soldesk\py_workspace
# 워크스페이스를 모듈 검색 경로에 추가
sys.path.insert(0, str(workspace))

# 등록된 경로 확인
print(sys.path[0])
```

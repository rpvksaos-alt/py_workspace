```python
''' 
1.클래스 선언
class 클래스명:
    def __init__(): -> 초기화 함수
        변수 선언

    함수 선언

2.객체 생성
클래스로 인스턴스 생성(메모리에 생성) -> 사용
변수 = 클래스명()
'''
```

```python
# 축구 선수 - 이름 , 백넘버, 나이 | 백넘버 변경하기...
class SoccerPlayer:
    # 속성 : init함수 안에 선언하는걸 추천
    # self : 현재 (생성하는) 인스턴스
    # __init__ 인스턴스 생성 시 제일 처음 1번 수행
    def __init__(self, n:str, back_n):
        self.name = n
        self.back_number = back_n
        self.age = 20
    #기능 - 함수로 선언. 클래스 안에 선언된 함수의 첫번째 매개변수 self!!
    def change_back_number(self, new_n):
        print(f'백넘버 {self.back_number}를 {new_n}로 변경합니다.')
        self.back_number = new_n

    def passing(self):
        print(f'{self.name}선수가 패스합니다.')
```

```python
# 인스턴스 생성
son = SoccerPlayer('흥민',7) # init에 명시한 매개변수에 들어갈 값 지정(self 제외)
print(son.name)
print(son.back_number)

#하나의 클래스로 여러개의 인스턴스를 만들 수 있다.
kim = SoccerPlayer('민재',3)
print(kim.name)
print(kim.back_number)

# 함수 호출 시 self를 제외한 나머지 매개변수에 값을 넣어줘야함
kim.passing()
kim.change_back_number(101)
print(kim.back_number)
print(son.back_number)
```

```python
# 학생(Student) - name, age | study
# 이름과 나이를 저장 할 수 있게 init함수 작성
# study 함수에서는 "'학생이름'이 열심히 공부하고 있습니다." 출력 
# 이름은 김스캇 나이는 10살인  학생 인스턴스 생성
# 공부시키기
class Student:
    def __init__(self,n1,a1):
        self.name = n1
        self.age = a1

    def study(self):
        print(f'{self.name}이 열심히 공부하고 있습니다.')
        

student1 = Student('김스캇',10)
print(student1.name)
print(student1.age)

student1.study()
```

```python
# 객체를 사용하는 이유
# - 관련된 데이터와 기능들을 하나로 묶어서 편하게 관리하기 위해서

# 객체를 사용하지 않고 학생 3명 선언
sn1 = 'scott'
sa1 = 10

sn2 = 'jane'
sa2 = 20

sn3 = 'kim'
sa3 = 21
```

```python
''' 
객체 사용 시 장점
1. 관련된 데이터를 묶을 수 있다.
2. 같은 구조의 데이터를 쉽게 만들 수 있다.
3. 데이터와 기능을 함께 관리할 수 있다.
'''
```

```python
''' 
def __init__(self, 초기 데이터) : 객체가 생성될 때 초기 데이터 받아서 저장

def __str__(self): 객체를 print() 했을 때 자동 호출. 반환 값이 출력된다.
                   객체에 저장된 값(정보)을 하나의 문자열로 만들어서 반환.
                   이 함수를 작성하지 않으면 객체의 타입과 주소값이 출력된다.
'''
```

```python
class Student:
    def __init__(self,name, age):
        self.name = name # 인스턴스 변수 name에 전달받은 값(name-지역변수) 저장 
        self.age = age

    def study(self):
       print(f'{self.name}학생이 열심히 공부하고 있습니다.')

    #객체에 저장된 값을 하나의 문자열로 만들어 리턴
    def __str__(self):
        return f'학생의 정보 : 이름:{self.name}, 나이:{self.age}'
        

s1 = Student('김스캇',10)
s1.study()
print(s1) # 객체의 주소값 출력 -> __str__호출 - return값 출력
```

```python
# 차(Car)
# color : str
# model : str
# speed : int
# accel
# braek --원래 break의 기능이 있기 때문에 일부러 스펠링을 다르게
# stop

class Car():
    def __init__(self, c, m):
        self.color = c # 인스턴스 생성 시에 원하는 값 저장
        self.model = m
        self.speed = 0

    def accel(self, s):
        self.speed += s
    def braek(self, s):
        self.speed -= s
    def stop(self):
        self.speed = 0

    def __str__(self):
        return f'{self.model}의 현재 속도는 {self.speed}km/h입니다.'

# 빨강색('red') '아이오닉3'
c1 = Car('red','아이오닉3')
c1.accel(90)
print(c1)
c1.braek(60)
print(c1)
c1.stop()
print(c1)

# 검정색('black') 'GV90'
c2 = Car('black','GV90')
c2.accel(110)
print(c2)
c2.accel(30)
print(c2)
c2.stop()
print(c2)
```

```python
# 객체 생성 후 속성 추가 할 수 있음

#인스턴스가 생성된 뒤에 변수를 추가하거나 삭제 가능
c2.number = '123가1234'# 외부에서 동적으로 추가
print(c2.number)
# print(c1.number) # 에러
```

```python
# 인스턴스가 생성된 뒤에 변수를 추가하거나 삭제 가능
del c2.number
# print(c2.number) # 에러
```

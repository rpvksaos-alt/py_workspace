```python
# 딕셔너리 : 이름이 있는 데이터들의 집합(key - value)
# 선언 { key:value, key:value } -> dictionary
# {value, value, value} -> set
```

```python
dic1 = { 1:'a', 2:'b', 3:'c'}
print(dic1,type(dic1))
```

```python
# 값의 의미를 명확하게
dic2 = {'name':'kim','age':10,'job':'programmer'}
print(dic2)
# print(dic2[0]) # 인덱스 없음
# 자료구조에서 원하는 데이터 추출
# 리스트 [인덱스]
# 딕셔너리[키]
print(dic1[1]) # 1은 인덱스 번호가 아니라 key이다
print(dic2['name'])
```

```python
dic3 = dict()
print(dic3)
```

```python
# 검색
# 키를 사용해 연결된 vlaue 꺼내온다.
# 1.딕셔너리[키]방법
print(dic1[1]) # 1은 인덱스 번호가 아니라 key이다.
print(dic2['name'])

print(dic2['gender']) # 없는 키
```

```python
# 2 딕셔너리.get(키)
print(dic1.get(1))
print(dic2.get('name'))
print(dic2.get('gender')) # 없는 키 -> None
```

```python
# 딕셔너리.get(키, 기본값) 키가 없을 시 None이 아니라 기본값으로 대체
print(dic2.get('name','이름 없음'))
print(dic2.get('gender','성별 없음'))
```

```python
# 데이터 추가
print(dic2)
# 딕셔너리[없는 키] = 값
dic2['gender'] = 'male'
print(dic2)
```

```python
# 재할당 (수정)
# 딕셔너리[기존 키] = 값 # 값만 변경
dic2['name'] ='james'
print(dic2)

# 키는 중복될 수 없기 때문에 같은 키값인 경우 값만 변경(나중에 들어온 값이 남는다)
dic3 = {'age':10, 'age':20}
print(dic3)
```

```python
# 삭제
del dic2['gender']
print(dic2)
```

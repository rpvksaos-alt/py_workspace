```python
# 학생 정보 저장
# 학번 : 1001
# 이름 : 김종현
# 학과 : 'AI융합'

student = {'학번':1001, '이름':'김종현', '학과':'AI융합'}
print(student, type(student))
```

```python
# key값만 구하기
print(student.keys()) # dict_keys 타입으로 키만 꺼내옴
# dict_keys 타입 -> 사용하기 편한 list로 바꿔서 쓰기
print(list(student.keys()))
```

```python
# for
# for 변수 in 딕셔너리 -> key 값만 꺼내와서 변수에 하나씩 저장
for s_info in student:
    print(s_info, student[s_info])
    # print(student.get(s_info))
```

```python
# value값들만 구하기
print(student.values())
print(list(student.values()))
```

```python
# item(k-v 한쌍) 구하기 - tuple(키0,값1)
print(student.items())
print(list(student.items()))
```

```python
for item in student.items():
    print(item[0], item[1])

print() # 줄 바꿈
for k, v in student.items(): # (key, value)
    print(k,v)    # 변수 바로 사용
```

```python
# in : 키가 있는지 확인

print('이름' in student)

if '주소' not in student:
    student['주소'] = '서울시 종로구'

print(student)    
```

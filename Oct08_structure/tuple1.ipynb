```python
# tuple
# 선언
t1 = (10, 20, 30)
print(t1, type(t1))
```

```python
i2 = (10) # int

print(i2, type(i2))
```

```python
t2 = (10,) # tuple에 저장할 데이터가 1개라면 ,와 같이 선언
print(t2, type(t2), len(t2))
```

```python
t3 = 10,20,30 # 자동으로 tuple로 묶어서 tuple 주소값 저장(패킹)
print(t3, type(t3),len(t3))

a,b,c = t3 # tuple에 저장된 데이터를 풀어서 각각의 변수에 저장(언패킹)
print(a,b,c)
```

```python
et1 = () # 비어있는 tuple
et2 = tuple() # 다른 타입의 값(list)을 받아서 tuple로 만들어주는 기능
              # 값을 안주면 비어있는 tuple 생성

print(et1, type(et1))
print(et2, type(et2))
```

```python
t4 = tuple((range(10)))
print(t4)
```

```python
# tuple -> 시퀀스 데이터 -> 인덱스가 붙어있음
# tuple 슬라이싱 -> tuple
print(t1[0])
print(t1[:2])
print(t1[::-1])
```

```python
for t in t1:
    print(t)
```

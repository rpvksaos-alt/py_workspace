```python
# 여러 클래스 간의 공통 멤버 -> 부모 클래스 선언
# 공통 멤버는 물려 받고 다른 멤버는 직접 선언
 
# 부모 클래스
class Mammal:
    def __init__(self, name):
        self.name = name
    def eat(self):
        print('밥 먹음')
    def sound(self):
        print('소리가 남')

class Dog(Mammal): # 상속. init, eat, sound 다 상속받음
    # 상속받은 함수를 현재 클래스에 맞게 재정의 -> 오버라이딩
    # 이름 변경 x, 수행 부분만 변경
    def sound(self):
        print('멍멍')

class Person(Mammal):
    def sound(self):
        print('말을 한다')
    def study(self):
        print('공부는 평생')

d1 = Dog('바둑이')
d1.eat()        
d1.sound() # 재정의된 함수 실행

p1 = Person('김철수')
p1.eat() # 물려받은 멤버
p1.study() # 직접 선언한 멤버
p1.sound() # 재정의된 함수 실행
```

```python
class Person:
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender

    def about_me(self):
          print(f'이름:{self.name}, 나이:{self.age}, 성별:{self.gender}')

# 부서(dept), 연봉(sal)
class Employee(Person):
       # 현재 클래스에 맞게 다시 선언
       def __init__(self,name,age,gender,dept,sal):
               super().__init__(name,age,gender) # super() : 부모 클래스. | 부모에 선언된 멤버는 부모 클래스에서 처리
               self.dept = dept
               self.sal = sal

       def do_work(self):
            print('일을 열심히 합니다.')

# 전공(major), 성적(grade)
# 성적 공개(show_info): 전공, 이름, 성적 출력
class Student(Person):
       def __init__(self,name,age,gender,major,grade):
            super().__init__(name,age,gender)
            self.major = major
            self.grade = grade

       def show_info(self):
            print(f'전공:{self.major}, 이름:{self.name}, 성적:{self.grade}')

# 'kim' / 30세 / '남' / '개발팀' / 5000
# 'smith' / 20세 / '남' / '철학' / 95         

e1 = Employee('kim', 30, '남', '개발팀', 5000)
print(e1.name)
e1.about_me()
e1.do_work()  

s1 = Student('smith', 20, '남', '철학', 95)
print(s1.age)
s1.show_info()
```

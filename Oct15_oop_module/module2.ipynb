```python
now = datetime. now()
# 날짜 중 원하는 값 추출
print(now.year)
print(now.month)
print(now.day)
print(now.minute)
print(now.second)
print(now.weekday()) # 요일 (월요일 0 ~ 일요일 6)
```

```python
# 특정날짜를 저장하기
my_birth = date(2000,1,1)
print(my_birth)

meeting_time = datetime(2026,12,24,15,30)
print(meeting_time)

# 날짜 - 날짜 : 날짜간의 차이를 일단위로 구함
print(meeting_time - datetime.now())
```

```python
# timedelta
from datetime import date , timedelta
today = date.today()
tomorrow = today + timedelta(days=1)
yesterday = today - timedelta(days=1)
next_week = today + timedelta(weeks=1)

print('오늘:', today)
print('내일:', tomorrow)
print('어제:', yesterday)
print('일주일 후:', next_week)
```

```python
# 날짜 형식 변경하기
# 날짜.strftime ('서식') : str <- time : 날짜값을 지정한 서식의 문자열로 바꿔줌(날짜를 원하는 형식으로 출력 시 사용)
# datetime.strptime ('날짜형식 문자열', '서식') : str -> time : 날짜 형식의 문자열을 날짜 데이터로 바꿔줌
# %Y %y %m %d %H %M %S %A

now = datetime.now()
print(now.strftime('%Y-%m-%d'))
print(now.strftime('%Y년-%m월-%d일 %A'))
print(now.strftime('%Y년-%m월-%d일 %A %H: %M: %S'))
```

```python
date_str = '2026년 10월 15일 2시 30분'
date_obj = datetime.strptime( date_str , '%Y년 %m월 %d일 %H시 %M분')
print(date_obj)
print(date_obj.hour)
```

```python
from datetime import date , timedelta
today = datetime.now()

# 1.오늘 날짜를 '26년 10월 15일' 형식으로 출력
print(today.strftime('%y년 %m월 %d일'))

# 오늘이 무슨 요일인지 한글로 출력
w = ['월','화','수','목','금','토','일']
print(w[today.weekday()])

# 오늘부터 100일 후 날짜 출력
print(today+timedelta(days=100))
```

```python
# 캡슐화 - 외부에서 객체의 정보(변수)에 직접 접근하지 못하게 감추기
# 객체의 내부 데이터를 직접 다루지 못하도록 감추고 -> __변수명 : 같은 클래스 내에서만 사용 가능
# 같은 클래스 안에 선언된 메서드를 통해 안전하게 접근하거나 변경하도록 하는 것
```

```python
class Account:
    def __init__(self, b):
        self.__balance = b # 캡슐화 적용 -> 이름을 변경해버림 (_클래스명__변수명)
                           # _Account__balance 로 변경됐음
    
    def deposit(self, money):
        if money > 0: # 함수 사용 시 데이터 보호 가능
            self.__balance += money

    def show_balance(self):
        print(f"잔액: {self.__balance}원")


account = Account(10000)
account.deposit(5000)
account.show_balance()

# print(account.__balance)

account.__balance = 'A' # 외부에서 직접 접근(잔액에 이상한 값을 넣을 수 있음) / __를 사용하니 접근 불가
                        #__balance라는 새로운 속성이 추가됨
                        # __balance(새로 추가된 속성), _Account__balance(원래 있던 속성-잔액)
account.show_balance()

print(account.__balance)
print(account._Account__balance) # 직접 접근할 수는 있지만 그렇게 사용하지 않음
```
